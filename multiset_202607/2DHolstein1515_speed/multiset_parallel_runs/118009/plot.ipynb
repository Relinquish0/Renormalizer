```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.animation import FuncAnimation
from IPython.display import HTML


# ==============================
# 1. 读取 Excel
# ==============================

file_path = "H1515_2D_m64_local_500steps_118009.xlsx"

# 假设 Excel 的全部 501 行都是数据
df = pd.read_excel(file_path, header=None)

print("Excel shape:", df.shape)


# ==============================
# 2. 转换为 numpy array
# ==============================

data = df.to_numpy(dtype=float)

n_frames, n_columns = data.shape

if n_columns != 225:
    raise ValueError(
        f"期望得到 225 列（15×15），"
        f"但现在得到 {n_columns} 列。"
    )


# ==============================
# 3. reshape
# ==============================

population = data.reshape(n_frames, 15, 15)

print("Population shape:", population.shape)


# ==============================
# 4. 全局颜色范围
# ==============================

vmin = np.nanmin(population)
vmax = np.nanmax(population)

print("Population min:", vmin)
print("Population max:", vmax)


# ==============================
# 5. 创建图像
# ==============================

fig, ax = plt.subplots(figsize=(6, 5))

im = ax.imshow(
    population[0],
    origin="lower",
    cmap="gray",
    vmin=vmin,
    vmax=vmax,
    interpolation="nearest",
    aspect="equal"
)

cbar = fig.colorbar(im, ax=ax)
cbar.set_label("Population")

ax.set_xlabel("X")
ax.set_ylabel("Y")

title = ax.set_title("Frame 0")


# ==============================
# 6. 更新动画
# ==============================

def update(frame):

    im.set_data(population[frame])

    # 不使用真实时间，只显示 frame
    title.set_text(
        f"Population | Frame {frame}/{n_frames - 1}"
    )

    return im, title


# ==============================
# 7. 创建动画
# ==============================

anim = FuncAnimation(
    fig,
    update,
    frames=n_frames,
    interval=50,       # ms/frame
    blit=False,
    repeat=True
)

plt.close(fig)


# ==============================
# 8. 在 Jupyter 中显示
# ==============================

HTML(anim.to_jshtml())
```
