```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

data = np.load("27x27.npz")
t, rho = data["time_series"], data["e_occupations"].reshape(-1, 27, 27)

@interact(step=IntSlider(0, 0, len(t) - 1, 1, description="step", continuous_update=True))
def show(step):
    plt.figure(figsize=(5, 4))
    plt.imshow(rho[step], origin="lower", cmap="gray_r", vmin=0, vmax=1)
    plt.colorbar(label=r"Exciton density $\rho$")
    plt.title(f"2D exciton diffusion: step {step}/{len(t)-1}, t={t[step]:.1f}")
    plt.xlabel("x"); plt.ylabel("y"); plt.show()
```
