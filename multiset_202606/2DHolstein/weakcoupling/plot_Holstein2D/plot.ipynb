```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import re

data_dir = Path("multiset_202606/2DHolstein/weakcoupling/plot_Holstein2D")
if not data_dir.exists():
    data_dir = Path(".")

files = sorted(data_dir.glob("*.npz"))
colors = {}
plt.figure()


for path in files:
    with np.load(path) as data:
        time = data["time_series"]
        entropy = data["S_maxbond"] if "S_maxbond" in data else np.max(np.real_if_close(data["bond_vn_entropy"]), axis=1)

    system = path.stem.split("_2D_")[0] if "_2D_" in path.stem else path.stem.split("_")[0]
    colors.setdefault(system, f"C{len(colors)}")
    ls = "-" if ("multiset" in path.stem or re.search(r"(?:_2D)?_m\d+\Z", path.stem)) else "--"
    plt.plot(time, entropy, ls=ls, color=colors[system], label=path.stem)

plt.xlabel("time")
plt.ylabel("max bond entropy")
plt.legend()
plt.tight_layout()
```

```python
plt.figure()

for path in files:
    with np.load(path) as data:
        eph_entropy = data["S_el"] if "S_el" in data else np.real_if_close(data["eph_vn_entropy"])
        max_bond_entropy = data["S_maxbond"] if "S_maxbond" in data else np.max(np.real_if_close(data["bond_vn_entropy"]), axis=1)

    system = path.stem.split("_2D_")[0] if "_2D_" in path.stem else path.stem.split("_")[0]
    ls = "-" if ("multiset" in path.stem or re.search(r"(?:_2D)?_m\d+\Z", path.stem)) else "--"
    plt.plot(eph_entropy, max_bond_entropy, ls=ls, color=colors[system], label=path.stem)

plt.xlabel("electron-phonon entropy")
plt.ylabel("max bond entropy")
plt.legend()
plt.tight_layout()
```
