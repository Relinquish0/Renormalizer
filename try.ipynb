```python
from renormalizer.model import HolsteinModel, Mol, Phonon  
from renormalizer import Mps, Quantity
  
# 创建声子（频率为100和200，假设使用cm^-1单位）  
# 第一个声子，频率100，物理维度设为8  
ph1 = Phonon.simple_phonon(Quantity(100, "cm^{-1}"), Quantity(1, "a.u."), 8)  
# 第二个声子，频率200，物理维度设为8    
ph2 = Phonon.simple_phonon(Quantity(200, "cm^{-1}"), Quantity(1, "a.u."), 8)  
```

```python
mol = Mol(Quantity(0, "a.u."), [ph1, ph2])  
model = HolsteinModel([mol, mol], Quantity(1, "eV"))  
mps = Mps.ground_state(model, max_entangled=False) 

```
